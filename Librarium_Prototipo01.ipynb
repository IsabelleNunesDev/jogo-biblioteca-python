```python
def evento_aquisicao(orcamento, acervo, instituicao, usuarios, equipe):

    print()
    print("Um professor solicitou novos livros para a biblioteca.")
    print()
    print("1 - Comprar todos os livros")
    print("2 - Comprar apenas os títulos prioritários")
    print("3 - Recusar a solicitação")

    escolha = input("Digite sua escolha: ")

    if escolha == "1":
        print("Você decidiu comprar todos os livros.")

        orcamento -= 15
        acervo += 15
        usuarios += 8

    elif escolha == "2":
        print("Você decidiu comprar apenas os títulos prioritários.")

        orcamento -= 8
        acervo += 8
        usuarios += 4

    elif escolha == "3":
        print("Você decidiu não realizar a aquisição.")

        orcamento += 2
        acervo -= 2
        usuarios -= 8
        instituicao -= 2

    else:
        print("Escolha inválida.")

    return orcamento, acervo, instituicao, usuarios, equipe

def evento_equipe(orcamento, acervo, instituicao, usuarios, equipe):

    print()
    print("A equipe informa que está sobrecarregada devido ao aumento do movimento.")
    print()
    print("1 - Manter o horário normal")
    print("2 - Reduzir temporariamente o horário")
    print("3 - Solicitar contratação temporária")

    escolha = input("Digite sua escolha: ")

    if escolha == "1":
        print("Você decidiu manter o horário normal.")

        equipe -= 10
        usuarios += 5

    elif escolha == "2":
        print("Você decidiu reduzir temporariamente o horário.")

        equipe += 8
        usuarios -= 10
        instituicao -= 2

    elif escolha == "3":
        print("Você decidiu solicitar uma contratação temporária.")

        orcamento -= 12
        equipe += 15
        usuarios += 5
        instituicao -= 3

    else:
        print("Escolha inválida.")

    return orcamento, acervo, instituicao, usuarios, equipe

def evento_instituicao(orcamento, acervo, instituicao, usuarios, equipe):

    print()
    print("A direção da instituição solicitou uma reunião. Segundo ela, os gastos da biblioteca aumentaram e é necessário demonstrar que os recursos estão sendo utilizados adequadamente.")
    print()
    print("1 - Apresentar um relatório detalhado")
    print("2 - Solicitar mais recursos")
    print("3 - Reduzir imediatamente os gastos")

    escolha = input("Digite sua escolha: ")

    if escolha == "1":
        print("Você decidiu apresentar um relatório detalhado.")

        equipe -= 5
        instituicao += 12
        orcamento += 3

    elif escolha == "2":
        print("Você decidiu solicitar mais recursos.")

        orcamento += 10
        usuarios += 3
        instituicao += 5

    elif escolha == "3":
        print("Você decidiu reduzir imediatamente os gastos.")

        orcamento += 10
        acervo -= 5
        usuarios -= 8
        instituicao += 3

    else:
        print("Escolha inválida.")

    return orcamento, acervo, instituicao, usuarios, equipe

orcamento = 50
acervo = 50
equipe = 50
instituicao = 50
usuarios = 50

rodada = 1

while rodada <=3:

    print()
    print("==============================")
    print("RODADA", rodada)
    print("==============================")

    print("1 - Evento de aquisição")
    print("2 - Evento da equipe")
    print("3 - Evento institucional")

    evento = input("Escolha o evento: ")

    if evento == "1":

        orcamento, acervo, instituicao, usuarios, equipe = evento_aquisicao(
            orcamento,
            acervo,
            instituicao,
            usuarios,
            equipe
        )

    elif evento == "2":

        orcamento, acervo, instituicao, usuarios, equipe = evento_equipe(
            orcamento,
            acervo,
            instituicao,
            usuarios,
            equipe
        )

    elif evento == "3":

        orcamento, acervo, instituicao, usuarios, equipe = evento_instituicao(
            orcamento,
            acervo,
            instituicao,
            usuarios,
            equipe
        )

    else:
        print("Evento inválido.")

    rodada += 1

print()
print("Estado da biblioteca:")
print("Orçamento =", orcamento)
print("Acervo =", acervo)
print("Instituição =", instituicao)
print("Usuários =", usuarios)
print("Equipe =", equipe)
```
